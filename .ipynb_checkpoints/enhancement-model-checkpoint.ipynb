# Image Analysis

```python
def analyze(img: np.ndarray) -> dict:
    """
    Compute quality metrics for one image.
    Used to decide which filters to apply.
    """
    gray   = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)          # uint8
    gray_f = gray.astype(np.float32)                        # float32 for math

    brightness = float(np.mean(gray_f))
    contrast   = float(gray_f.std())
    sharpness  = float(cv2.Laplacian(gray, cv2.CV_32F).var())  # uint8 -> CV_32F
    dark_ratio = float(np.sum(gray < 50) / gray.size)

    local_mean = cv2.blur(gray_f, (5, 5))
    noise_std  = float(np.std(np.abs(gray_f - local_mean)))
    snr        = brightness / (noise_std + 1e-6)

    return {
        'brightness'      : brightness,
        'contrast'        : contrast,
        'sharpness'       : sharpness,
        'dark_ratio'      : dark_ratio,
        'snr'             : snr,
        # flags
        'is_dark'         : brightness < 80,
        'is_low_contrast' : contrast   < 40,
        'is_blurry'       : sharpness  < 150,
        'is_noisy'        : snr        < 5.0,
        'has_dark_regions': dark_ratio > 0.15,
    }
```

# Filters

```python
def clahe(img: np.ndarray, clip: float = 3.0) -> np.ndarray:
    """
    Contrast-Limited Adaptive Histogram Equalization on L channel (LAB).
    Improves local contrast without colour distortion.
    Essential for aerial imagery.
    """
    c       = cv2.createCLAHE(clipLimit=clip, tileGridSize=(8, 8))
    lab     = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)
    l, a, b = cv2.split(lab)
    return cv2.cvtColor(cv2.merge([c.apply(l), a, b]), cv2.COLOR_LAB2BGR)


def gamma_correct(img: np.ndarray, gamma: float = 1.5) -> np.ndarray:
    """
    Gamma correction via LUT.
    gamma > 1 → brightens darks.
    gamma < 1 → dims over-exposed areas.
    Preserves colour ratios between pixels.
    """
    table = np.array(
        [(i / 255.0) ** (1.0 / gamma) * 255 for i in range(256)],
        dtype=np.uint8
    )
    return cv2.LUT(img, table)


def shadow_recovery(img: np.ndarray, strength: int = 60) -> np.ndarray:
    """
    Lifts shadow regions only (V < 100 in HSV).
    Does not affect highlights — mimics Lightroom shadow slider.
    """
    hsv     = cv2.cvtColor(img, cv2.COLOR_BGR2HSV).astype(np.float32)
    h, s, v = cv2.split(hsv)
    mask    = (v < 100).astype(np.float32)
    v       = np.clip(v + strength * mask * (1.0 - v / 100.0), 0, 255)
    return cv2.cvtColor(cv2.merge([h, s, v]).astype(np.uint8), cv2.COLOR_HSV2BGR)


def denoise(img: np.ndarray) -> np.ndarray:
    """
    Non-Local Means denoising.
    Removes sensor noise while preserving edges.
    Must run BEFORE sharpening to avoid amplifying noise.
    """
    return cv2.fastNlMeansDenoisingColored(
        img, None,
        h=8, hColor=8,
        templateWindowSize=7,
        searchWindowSize=21
    )


def unsharp_mask(img: np.ndarray, sigma: float = 1.5,
                 strength: float = 1.3) -> np.ndarray:
    """
    Unsharp masking: subtracts a blurred version to enhance edges.
    Best sharpening method for satellite imagery.
    """
    blurred = cv2.GaussianBlur(img, (0, 0), sigma)
    sharp   = cv2.addWeighted(img, 1.0 + strength, blurred, -strength, 0)
    return np.clip(sharp, 0, 255).astype(np.uint8)


def bilateral(img: np.ndarray) -> np.ndarray:
    """
    Edge-preserving smoothing.
    Reduces fine noise without blurring object boundaries.
    Run before unsharp mask.
    """
    return cv2.bilateralFilter(img, d=9, sigmaColor=75, sigmaSpace=75)


def detail_enhance(img: np.ndarray) -> np.ndarray:
    """
    OpenCV detail enhancement.
    Sharpens fine textures — helps detect small objects (vehicles, planes).
    """
    return cv2.detailEnhance(img, sigma_s=10, sigma_r=0.12)


def hist_stretch(img: np.ndarray) -> np.ndarray:
    """
    Per-channel linear histogram stretching.
    Expands the dynamic range when contrast is very low.
    """
    out = np.zeros_like(img)
    for i in range(3):
        ch = img[:, :, i].astype(np.float32)
        lo, hi = ch.min(), ch.max()
        out[:, :, i] = (
            ((ch - lo) / (hi - lo) * 255).astype(np.uint8) if hi > lo
            else img[:, :, i]
        )
    return out


def color_boost(img: np.ndarray, scale: float = 1.15) -> np.ndarray:
    """
    Mild saturation boost to improve object-background separation.
    Kept subtle (1.15) to avoid disturbing detection model colour space.
    """
    hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV).astype(np.float32)
    hsv[:, :, 1] = np.clip(hsv[:, :, 1] * scale, 0, 255)
    return cv2.cvtColor(hsv.astype(np.uint8), cv2.COLOR_HSV2BGR)
```

# Smart Enhancement Pipeline

```python
def enhance_image(img: np.ndarray) -> np.ndarray:
    """
    Standalone enhancement function.

    Accepts a BGR numpy array (as returned by cv2.imread).
    Returns an enhanced BGR numpy array of the same size.

    Pipeline (adaptive — filters only applied when needed):
        1. Denoise          — always if noisy
        2. Shadow Recovery  — if dark regions detected
        3. Gamma Correction — if overall dark
        4. Hist Stretch     — if contrast is very low
        5. CLAHE            — always (core enhancement)
        6. Bilateral        — always (pre-sharpening smooth)
        7. Unsharp Mask     — if blurry or low contrast
        8. Detail Enhance   — always
        9. Color Boost      — always (mild)

    This function is self-contained and has no dependency on any
    other model or external state. The backend can import and call it
    directly on any satellite image array.
    """
    if img is None or img.size == 0:
        raise ValueError("enhance_image received an empty or None image.")

    info = analyze(img)
    out  = img.copy()

    # 1 — Denoise
    if info['is_noisy'] or info['contrast'] < 50:
        out = denoise(out)

    # 2 — Shadow Recovery
    if info['has_dark_regions'] or info['is_dark']:
        strength = 80 if info['brightness'] < 50 else 55
        out = shadow_recovery(out, strength=strength)

    # 3 — Gamma
    if info['is_dark']:
        g = 1.8 if info['brightness'] < 50 else 1.4
        out = gamma_correct(out, gamma=g)

    # 4 — Histogram Stretch
    if info['is_low_contrast']:
        out = hist_stretch(out)

    # 5 — CLAHE (core step — always)
    clip = 4.0 if info['is_low_contrast'] else 2.5
    out  = clahe(out, clip=clip)

    # 6 — Bilateral
    out = bilateral(out)

    # 7 — Unsharp Mask
    if info['is_blurry'] or info['contrast'] < 60:
        s = 1.5 if info['sharpness'] < 80 else 1.1
        out = unsharp_mask(out, strength=s)

    # 8 — Detail Enhancement
    out = detail_enhance(out)

    # 9 — Color Boost
    out = color_boost(out)

    return out
```

# Validate on Single Image

```python
sample_path = None
sample_orig = None
for _fname in sorted(os.listdir(IMG_DIR)):
    if Path(_fname).suffix.lower() not in SUPPORTED_EXT:
        continue
    _candidate = os.path.join(IMG_DIR, _fname)
    _img = cv2.imread(_candidate)
    if _img is not None and _img.size > 0:
        sample_path = _candidate
        sample_orig = _img
        print(f"Sample image : {_fname}  |  shape: {_img.shape}")
        break

if sample_orig is None:
    raise FileNotFoundError(f"No readable images found in {IMG_DIR}. Check IMG_DIR path.")

sample_enh = enhance_image(sample_orig)

before = analyze(sample_orig)
after  = analyze(sample_enh)

print("── Single-image validation ────────────────────────────────────────────")
print(f"  Image      : {Path(sample_path).name}")
print(f"  Brightness : {before['brightness']:.1f}  →  {after['brightness']:.1f}")
print(f"  Contrast   : {before['contrast']:.1f}  →  {after['contrast']:.1f}")
print(f"  Sharpness  : {before['sharpness']:.1f}  →  {after['sharpness']:.1f}")

fig, axes = plt.subplots(1, 2, figsize=(14, 6))
axes[0].imshow(cv2.cvtColor(sample_orig, cv2.COLOR_BGR2RGB))
axes[0].set_title(f"Original\nbright={before['brightness']:.0f} | "
                  f"contrast={before['contrast']:.0f} | sharp={before['sharpness']:.0f}",
                  fontsize=11)
axes[0].axis('off')

axes[1].imshow(cv2.cvtColor(sample_enh, cv2.COLOR_BGR2RGB))
axes[1].set_title(f"Enhanced\nbright={after['brightness']:.0f} | "
                  f"contrast={after['contrast']:.0f} | sharp={after['sharpness']:.0f}",
                  fontsize=11)
axes[1].axis('off')

fig.suptitle("Single-Image Validation", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig(os.path.join(REPORT_DIR, 'single_validation.png'), dpi=150, bbox_inches='tight')
plt.show()
```

# Filter Showcase (step-by-step on sample image)

```python
steps = {
    'Original'       : sample_orig,
    'Denoise'        : denoise(sample_orig),
    'Shadow Recovery': shadow_recovery(sample_orig, 65),
    'Gamma (1.5)'    : gamma_correct(sample_orig, 1.5),
    'Hist Stretch'   : hist_stretch(sample_orig),
    'CLAHE'          : clahe(sample_orig, clip=3.0),
    'Bilateral'      : bilateral(sample_orig),
    'Unsharp Mask'   : unsharp_mask(sample_orig),
    'Detail Enhance' : detail_enhance(sample_orig),
    'Color Boost'    : color_boost(sample_orig),
    'Full Pipeline'  : sample_enh,
}

cols = 4
rows = -(-len(steps) // cols)
fig, axes = plt.subplots(rows, cols, figsize=(6 * cols, 5 * rows))
axes = axes.flatten()

for ax, (name, fimg) in zip(axes, steps.items()):
    info = analyze(fimg)
    ax.imshow(cv2.cvtColor(fimg, cv2.COLOR_BGR2RGB))
    ax.set_title(
        f"{name}\n"
        f"B={info['brightness']:.0f} | C={info['contrast']:.0f} | S={info['sharpness']:.0f}",
        fontsize=9, fontweight='bold'
    )
    ax.axis('off')

for ax in axes[len(steps):]:
    ax.set_visible(False)

fig.suptitle("Filter-by-Filter Breakdown", fontsize=16, fontweight='bold')
plt.tight_layout()
plt.savefig(os.path.join(REPORT_DIR, 'filter_showcase.png'), dpi=150, bbox_inches='tight')
plt.show()
```

# Benchmark on Sample Images

```python
BENCHMARK_N = 100

all_files = sorted([
    f for f in os.listdir(IMG_DIR)
    if Path(f).suffix.lower() in SUPPORTED_EXT
])

# Evenly spaced sample across the dataset
step         = max(1, len(all_files) // BENCHMARK_N)
sample_files = all_files[::step][:BENCHMARK_N]
print(f"\nBenchmarking on {len(sample_files)} / {len(all_files)} images...")

records = []
failed  = []

for idx, fname in enumerate(sample_files, 1):
    src = os.path.join(IMG_DIR, fname)
    img = cv2.imread(src)

    if img is None:
        failed.append(fname)
        continue

    # Resize large images for speed (enhancement logic is resolution-agnostic)
    h, w = img.shape[:2]
    if max(h, w) > 1024:
        scale = 1024 / max(h, w)
        img   = cv2.resize(img, (int(w*scale), int(h*scale)))

    before_info = analyze(img)

    try:
        enhanced = enhance_image(img)
    except Exception as e:
        print(f"  [ERROR] {fname}: {e}")
        failed.append(fname)
        continue

    after_info = analyze(enhanced)
    cv2.imwrite(os.path.join(OUTPUT_DIR, fname), enhanced)

    records.append({
        'image'            : fname,
        'brightness_before': round(before_info['brightness'], 2),
        'brightness_after' : round(after_info['brightness'],  2),
        'brightness_gain'  : round(after_info['brightness']  - before_info['brightness'],  2),
        'contrast_before'  : round(before_info['contrast'],   2),
        'contrast_after'   : round(after_info['contrast'],    2),
        'contrast_gain'    : round(after_info['contrast']    - before_info['contrast'],    2),
        'sharpness_before' : round(before_info['sharpness'],  2),
        'sharpness_after'  : round(after_info['sharpness'],   2),
        'sharpness_gain'   : round(after_info['sharpness']   - before_info['sharpness'],   2),
    })
    print(f"  [{idx}/{len(sample_files)}] {fname}")

print(f"\n✓ Benchmarked : {len(records)} images")
print(f"✗ Failed       : {len(failed)} images")
```

# Stats Report

```python
df = pd.DataFrame(records)
df.to_csv(os.path.join(REPORT_DIR, 'enhancement_stats.csv'), index=False)

print("\n── Enhancement Gains (mean across all images) ─────────────────────────")
print(df[['brightness_gain', 'contrast_gain', 'sharpness_gain']].mean().round(2))

fig, axes = plt.subplots(1, 3, figsize=(17, 4))
metrics = [
    ('brightness_gain', 'Brightness Gain', '#4C72B0'),
    ('contrast_gain',   'Contrast Gain',   '#DD8452'),
    ('sharpness_gain',  'Sharpness Gain',  '#55A868'),
]
for ax, (col, title, color) in zip(axes, metrics):
    ax.hist(df[col], bins=25, color=color, edgecolor='white', alpha=0.85)
    ax.axvline(0,    color='red', linestyle='--', linewidth=1.5, label='No change')
    ax.axvline(df[col].mean(), color='black', linestyle='-',
               linewidth=1.5, label=f"Mean={df[col].mean():.1f}")
    ax.set_title(title, fontsize=13, fontweight='bold')
    ax.set_xlabel('Gain (After − Before)')
    ax.set_ylabel('# Images')
    ax.legend(); ax.grid(axis='y', alpha=0.3)

fig.suptitle("Enhancement Gains Across Dataset", fontsize=15, fontweight='bold')
plt.tight_layout()
plt.savefig(os.path.join(REPORT_DIR, 'gains_distribution.png'), dpi=150, bbox_inches='tight')
plt.show()
```

# Save Model as Python Module

```python
module_code = '''"""
enhancement_model.py
====================
Standalone satellite image enhancement model.

Usage
-----
    import cv2
    from enhancement_model import enhance_image

    img      = cv2.imread("path/to/image.png")   # BGR numpy array
    enhanced = enhance_image(img)                 # BGR numpy array (same size)
    cv2.imwrite("enhanced.png", enhanced)
"""

import cv2
import numpy as np


def _analyze(img):
    gray       = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY).astype(np.float32)
    brightness = float(np.mean(gray))
    contrast   = float(gray.std())
    sharpness  = float(cv2.Laplacian(gray, cv2.CV_64F).var())
    dark_ratio = float(np.sum(gray < 50) / gray.size)
    local_mean = cv2.blur(gray, (5, 5))
    snr        = brightness / (float(np.std(np.abs(gray - local_mean))) + 1e-6)
    return {
        "brightness"      : brightness,
        "contrast"        : contrast,
        "sharpness"       : sharpness,
        "dark_ratio"      : dark_ratio,
        "is_dark"         : brightness < 80,
        "is_low_contrast" : contrast   < 40,
        "is_blurry"       : sharpness  < 150,
        "is_noisy"        : snr        < 5.0,
        "has_dark_regions": dark_ratio > 0.15,
    }


def enhance_image(img: np.ndarray) -> np.ndarray:
    """
    Enhance a satellite image.

    Parameters
    ----------
    img : np.ndarray
        BGR image array as returned by cv2.imread().

    Returns
    -------
    np.ndarray
        Enhanced BGR image array, same spatial dimensions as input.
    """
    if img is None or img.size == 0:
        raise ValueError("enhance_image received an empty or None image.")

    info = _analyze(img)
    out  = img.copy()

    # Denoise
    if info["is_noisy"] or info["contrast"] < 50:
        out = cv2.fastNlMeansDenoisingColored(out, None, 8, 8, 7, 21)

    # Shadow recovery
    if info["has_dark_regions"] or info["is_dark"]:
        strength = 80 if info["brightness"] < 50 else 55
        hsv      = cv2.cvtColor(out, cv2.COLOR_BGR2HSV).astype(np.float32)
        h, s, v  = cv2.split(hsv)
        mask     = (v < 100).astype(np.float32)
        v        = np.clip(v + strength * mask * (1.0 - v / 100.0), 0, 255)
        out      = cv2.cvtColor(cv2.merge([h, s, v]).astype(np.uint8), cv2.COLOR_HSV2BGR)

    # Gamma correction
    if info["is_dark"]:
        g     = 1.8 if info["brightness"] < 50 else 1.4
        table = np.array([(i/255.0)**(1.0/g)*255 for i in range(256)], np.uint8)
        out   = cv2.LUT(out, table)

    # Histogram stretch
    if info["is_low_contrast"]:
        tmp = np.zeros_like(out)
        for i in range(3):
            ch = out[:,:,i].astype(np.float32)
            lo, hi = ch.min(), ch.max()
            tmp[:,:,i] = ((ch-lo)/(hi-lo)*255).astype(np.uint8) if hi > lo else out[:,:,i]
        out = tmp

    # CLAHE
    clip  = 4.0 if info["is_low_contrast"] else 2.5
    c     = cv2.createCLAHE(clipLimit=clip, tileGridSize=(8, 8))
    lab   = cv2.cvtColor(out, cv2.COLOR_BGR2LAB)
    l,a,b = cv2.split(lab)
    out   = cv2.cvtColor(cv2.merge([c.apply(l), a, b]), cv2.COLOR_LAB2BGR)

    # Bilateral
    out = cv2.bilateralFilter(out, 9, 75, 75)

    # Unsharp mask
    if info["is_blurry"] or info["contrast"] < 60:
        s       = 1.5 if info["sharpness"] < 80 else 1.1
        blurred = cv2.GaussianBlur(out, (0,0), 1.5)
        out     = np.clip(cv2.addWeighted(out, 1+s, blurred, -s, 0), 0, 255).astype(np.uint8)

    # Detail enhance
    out = cv2.detailEnhance(out, sigma_s=10, sigma_r=0.12)

    # Color boost
    hsv          = cv2.cvtColor(out, cv2.COLOR_BGR2HSV).astype(np.float32)
    hsv[:,:,1]   = np.clip(hsv[:,:,1] * 1.15, 0, 255)
    out          = cv2.cvtColor(hsv.astype(np.uint8), cv2.COLOR_HSV2BGR)

    return out
'''

module_path = os.path.join(MODEL_DIR, 'enhancement_model.py')
with open(module_path, 'w') as f:
    f.write(module_code)

print(f"✓ Model saved → {module_path}")
print("\nBackend usage:")
print("  from enhancement_model import enhance_image")
print("  enhanced = enhance_image(cv2.imread('image.png'))")

print("\n" + "="*60)
print("Model (Enhancement) — COMPLETE")
print("="*60)
print(f"  Enhanced images : {OUTPUT_DIR}/")
print(f"  Reports         : {REPORT_DIR}/")
print(f"  Model file      : {module_path}")
```
